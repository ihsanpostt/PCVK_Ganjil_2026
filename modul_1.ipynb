```python

import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt


black_img = np.zeros(shape=(512,512,3), dtype=np.int16)
plt.imshow(black_img)
```

```python
# perhatikan koordinat titik2 pt1 dan pt2
cv.rectangle(black_img, pt1=(384,0), pt2=(510,150), color=(0,255,0), thickness=10)
plt.imshow(black_img)
```

```python
cv.rectangle(black_img, pt1=(200,200), pt2=(300,300), color=(0,0,255), thickness=15)
plt.imshow(black_img)

cv.circle(black_img, center=(100,100), radius=50, color=(255,0,0), thickness=8)
plt.imshow(black_img)
```

```python
cv.line(black_img, pt1=(0,0), pt2=(512,512), color=(255,255,255), thickness=5)
plt.imshow(black_img)

font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img, text='Hello', org=(10,500), fontFace=font, fontScale=4, color=(255,255,0), thickness=2, lineType=cv.LINE_AA)
plt.imshow(black_img)
```

```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt


black_img2 = np.zeros(shape=(512,512,3), dtype=np.int32)
plt.imshow(black_img2)


vertices = np.array([[100,300],[200,200],[400,300],[200,400]], dtype=np.int32)

pts = vertices.reshape((-1,1,2))
pts


cv.polylines(black_img2, [pts], isClosed=True, color=(255,0,0), thickness=5)
plt.imshow(black_img2)
```

```python
# Tampilkan Red-Blue (Nol-kan Green/index 1)
img_RB = imgRGB.copy()
img_RB[:, :, 1] = 0

# Tampilkan Green-Blue (Nol-kan Red/index 0)
img_GB = imgRGB.copy()
img_GB[:, :, 0] = 0

plt.subplot(1, 2, 1); plt.imshow(img_RB); plt.title("Red-Blue")
plt.subplot(1, 2, 2); plt.imshow(img_GB); plt.title("Green-Blue")
plt.show()
```

```python
img_red_only = imgRGB[5:30, :].copy()
img_red_only[:, :, 1] = 0 # Hilangkan Green
img_red_only[:, :, 2] = 0 # Hilangkan Blue
plt.imshow(img_red_only)
plt.title("Baris 5-30, Channel Red Saja")
plt.show()
```

```python
import random

img_kotak = imgRGB.copy()
h, w = img_kotak.shape[:2]

for _ in range(5):
    # Kordinat acak
    x1, y1 = random.randint(0, w-50), random.randint(0, h-50)
    x2, y2 = x1 + random.randint(20, 100), y1 + random.randint(20, 100)
    # Warna acak RGB
    r, g, b = random.randint(0, 255), random.randint(0, 255), random.randint(0, 255)

    cv2.rectangle(img_kotak, (x1, y1), (x2, y2), (r, g, b), 3)

plt.imshow(img_kotak)
plt.title("5 Kotak Random")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# 1. Baca gambar & konversi ke RGB
foto = cv2.imread("/content/drive/MyDrive/PCVK/female.jpg")
foto_RGB = cv2.cvtColor(foto, cv2.COLOR_BGR2RGB)
h, w, _ = foto_RGB.shape

# 2. Split channel & buat rectangle khusus di Channel B
R, G, B = cv2.split(foto_RGB)
cv2.rectangle(B, pt1=(10, h - 80), pt2=(240, h - 10), color=255, thickness=3)
foto_RGB = cv2.merge((R, G, B))

# 3. Tambahkan teks nama file (Warna Cyan & Font Duplex)
cv2.putText(foto_RGB, "female.webp", (25, h - 35), cv2.FONT_HERSHEY_DUPLEX, 0.7, (0, 255, 255), 2)

# 4. Tampilkan citra
plt.imshow(foto_RGB)
plt.axis("off")
plt.title("Tugas 8 dan 9")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# 1. Baca gambar KTM
ktm = cv2.imread("/content/drive/MyDrive/PCVK/ktm.jpeg")
ktm_RGB = cv2.cvtColor(ktm, cv2.COLOR_BGR2RGB)

h, w, _ = ktm_RGB.shape

# 2. Gambar Mata Kiri (Lingkaran Merah)
cv2.circle(ktm_RGB, center=(int(w * 0.116), int(h * 0.680)), radius=int(w * 0.009), color=(255, 0, 0), thickness=3)

# 3. Gambar Mata Kanan (Lingkaran Hijau)
cv2.circle(ktm_RGB, center=(int(w * 0.138), int(h * 0.680)), radius=int(w * 0.009), color=(0, 255, 0), thickness=3)

# 4. Gambar Mulut (Kotak Biru)
cv2.rectangle(ktm_RGB, pt1=(int(w * 0.112), int(h * 0.720)), pt2=(int(w * 0.142), int(h * 0.738)), color=(0, 0, 255), thickness=3)

# 5. Tampilkan Hasil
plt.imshow(ktm_RGB)
plt.show()
```
